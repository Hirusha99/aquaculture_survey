```python
from sklearn.preprocessing import StandardScaler

columns_to_standardize = [
'DURATIONMONTHS',
'INCOME',
'COST_TOTAL',
'AVG_WEIGHT_KG',
'PRODUCTION_HA'

]

scaler = StandardScaler()

df[columns_to_standardize] = scaler.fit_transform(
    df[columns_to_standardize]
)

df[columns_to_standardize].head()
```

```python
# Null / NaN count for each column
df.isnull().sum()
```

```python
import numpy as np
df[columns_to_standardize] = df[columns_to_standardize].replace(
    [np.inf, -np.inf],
    np.nan
)

df = df.dropna(subset=columns_to_standardize)
```

```python
# Null / NaN count for each column
df.isnull().sum()
```

```python
X = df[
    [

'INCOME',
'COST_TOTAL',
'AVG_WEIGHT_KG'

    ]
]

Y = df[
    [
        'DURATIONMONTHS',
        'PRODUCTION_HA'
    ]
]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

for i, target in enumerate(['Y1', 'Y2']):

    print(target)

    print(
        "RMSE:",
        np.sqrt(mean_squared_error(
            y_test.iloc[:, i],
            y_pred[:, i]
        ))
    )

    print(
        "MAE:",
        mean_absolute_error(
            y_test.iloc[:, i],
            y_pred[:, i]
        )
    )

    print(
        "R2:",
        r2_score(
            y_test.iloc[:, i],
            y_pred[:, i]
        )
    )

    print()
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=300,
    random_state=42
)

rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
```

```python
for i, target in enumerate(['Y1', 'Y2']):

    print(target)

    print(
        "RMSE:",
        np.sqrt(mean_squared_error(
            y_test.iloc[:, i],
            y_pred_rf[:, i]
        ))
    )

    print(
        "MAE:",
        mean_absolute_error(
            y_test.iloc[:, i],
            y_pred_rf[:, i]
        )
    )

    print(
        "R2:",
        r2_score(
            y_test.iloc[:, i],
            y_pred_rf[:, i]
        )
    )

    print()
```

```python
from sklearn.multioutput import MultiOutputRegressor
from xgboost import XGBRegressor

xgb = MultiOutputRegressor(
    XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=4,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    )
)

xgb.fit(X_train, y_train)

y_pred_xgb = xgb.predict(X_test)
```

```python
for i, target in enumerate(['Y1', 'Y2']):

    print(target)

    print(
        "RMSE:",
        np.sqrt(mean_squared_error(
            y_test.iloc[:, i],
            y_pred_xgb[:, i]
        ))
    )

    print(
        "MAE:",
        mean_absolute_error(
            y_test.iloc[:, i],
            y_pred_xgb[:, i]
        )
    )

    print(
        "R2:",
        r2_score(
            y_test.iloc[:, i],
            y_pred_xgb[:, i]
        )
    )

    print()
```

```python
model_y1 = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    random_state=42
)

model_y2 = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    random_state=42
)

model_y1.fit(X, y1)
model_y2.fit(X, y2)
```

```python
features = [
    'INCOME',
    'COST_TOTAL',
    'AVG_WEIGHT_KG'
]
```

```python
import numpy as np
import pandas as pd

n_samples = 70

candidate_data = pd.DataFrame()

for feature in features:
    candidate_data[feature] = np.random.uniform(
        df[feature].min(),
        df[feature].max(),
        n_samples
    )
```

```python
candidate_data['Predicted_Y1'] = model_y1.predict(
    candidate_data[features]
)

candidate_data['Predicted_Y2'] = model_y2.predict(
    candidate_data[features]
)
```

```python
candidate_data['Y1_norm'] = (
    candidate_data['Predicted_Y1']
    - candidate_data['Predicted_Y1'].min()
) / (
    candidate_data['Predicted_Y1'].max()
    - candidate_data['Predicted_Y1'].min()
)

candidate_data['Y2_norm'] = (
    candidate_data['Predicted_Y2']
    - candidate_data['Predicted_Y2'].min()
) / (
    candidate_data['Predicted_Y2'].max()
    - candidate_data['Predicted_Y2'].min()
)
```

```python
candidate_data['Optimization_Score'] = (
    0.5 * candidate_data['Y1_norm']
    + 0.5 * candidate_data['Y2_norm']
)
```

```python
new_input_scaled = scaler_new.transform(new_input[features])

pred_y1_scaled = model_y1.predict(new_input_scaled)[0]
pred_y2_scaled = model_y2.predict(new_input_scaled)[0]

# Identify the indices of 'PRODUCTION_HA' and 'DURATIONMONTHS' in the original scaler's features
production_ha_index = columns_to_standardize.index('PRODUCTION_HA')
durationmonths_index = columns_to_standardize.index('DURATIONMONTHS')

# Inverse transform pred_y1 (PRODUCTION_HA)
pred_y1_original = (pred_y1_scaled * scaler.scale_[production_ha_index]) + scaler.mean_[production_ha_index]

# Inverse transform pred_y2 (DURATIONMONTHS)
pred_y2_original = (pred_y2_scaled * scaler.scale_[durationmonths_index]) + scaler.mean_[durationmonths_index]

print("Predicted DURATIONMONTHS (Standardized):", pred_y2_scaled)
print("Predicted PRODUCTION_HA (Standardized):", pred_y1_scaled)

print("Predicted DURATIONMONTHS (Original Scale):", pred_y2_original)
print("Predicted PRODUCTION_HA (Original Scale):", pred_y1_original)
```
